# Import Libraries

```python
import matplotlib.pyplot as plt
```

# Create Dataset

```python
from sklearn.datasets import make_classification

X, y = make_classification(n_samples=1000, n_features=2, n_classes=2, n_redundant=0, random_state=0)
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y)
```

```python
# Sépare les sous-bases, une pour l'entraînement et une pour l'évaluation
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=101)
```

# Model 1

```python
from sklearn.ensemble import RandomForestClassifier

clf = RandomForestClassifier(random_state=0)
clf.fit(X_train, y_train)
```

```python
print(clf.get_params())
```

```python
from sklearn import tree

plt.figure(figsize=(20, 12))
tree.plot_tree(clf.estimators_[1], filled=True)
```

```python
print(clf.predict([[0, 0]]))
```

```python
y_pred = clf.predict(X_test)
```

```python
from sklearn import metrics
```

```python
print(metrics.classification_report(y_test, y_pred))
```

```python
confusion_matrix = metrics.confusion_matrix(y_test, y_pred)
cm_display = metrics.ConfusionMatrixDisplay(confusion_matrix = confusion_matrix)

cm_display.plot()
plt.show()
```
